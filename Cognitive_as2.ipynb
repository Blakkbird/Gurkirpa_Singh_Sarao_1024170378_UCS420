## Q1 - List L from the roll number digits

```python
L = [d * 10 for d in digits]
```

**Q1(i)**

```python
print("L =", L) 
```

**Q1(ii)**

```python
L.append(50)
print("After append(50):", L)      # append adds at the end

L.insert(3, 80)
print("After insert(3, 80):", L)   # insert puts 80 at index 3, rest shifts right
```

**Q1(iii)**

```python
L.remove(50)
print("After remove(50):", L)      # remove deletes by value

last = L.pop()
print("pop() returned:", last)     # pop deletes by position and returns the value
print("After pop():", L)
```

**Q1(iv)**

```python
L.sort()
print("Ascending:", L)

L.sort(reverse=True)
print("Descending:", L)
```

**Q1(v)**

```python
print("First three:", L[:3])
print("Last three:", L[-3:])
```

**Q1(vi)**

```python
avg = sum(L) / len(L)
print("Average =", avg)

above = [x for x in L if x > avg]
print("Above average:", above)
```

## Q2 - Tuple of scores

```python
scores = tuple(L[:8])
print("scores =", scores)
```

**Q2(i)**

```python
print("Highest score =", max(scores))
print("Its index =", scores.index(max(scores)))
print("Lowest score =", min(scores))
print("Times it appears =", scores.count(min(scores)))
```

```python
# Tuples are immutable, so there is no reverse() like a list has. Slicing makes a new one.
print("Reversed:", list(scores[::-1]))
```

**Q2(v)**

```python
first, second, *rest = scores
print("First =", first)
print("Second =", second)
print("Remaining =", rest)
```

## Q3 - 100 random numbers

```python
random.seed(roll)
```

**Q3(i)**

```python
nums = [random.randint(100, 900) for _ in range(100)]
print("Numbers:", nums)
```

**Q3(ii)**

```python
odd = [x for x in nums if x % 2 != 0]
print("Count of odd numbers =", len(odd))
print("Odd numbers:", odd)
```

**Q3(iii)**

```python
even = [x for x in nums if x % 2 == 0]
print("Count of even numbers =", len(even))
print("Even numbers:", even)
```

**Q3(iv)**

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True


primes = [x for x in nums if is_prime(x)]
print("Count of prime numbers =", len(primes))
print("Prime numbers:", primes)
```

**Q3(v)**

```python
count = Counter(nums)
num, freq = count.most_common(1)[0]
print("Most frequent number =", num)
print("It occurs", freq, "times")
```

## Q4 - Sets A and B

```python
d = digits[:8]

A = {x * 7 for x in d}
B = {x * 9 for x in d}

print("A =", A)
print("B =", B)
```

**Q4(vi)**

```python
print("A union B =", A.union(B))
```

**Q4(vii)**

```python
print("A intersection B =", A.intersection(B))
```

**Q4(viii)**

```python
print("A - B =", A.difference(B))
print("B - A =", B.difference(A))

# difference() is one directional. symmetric_difference() gives the leftovers of both sides together.
```

**Q4(ix)**

```python
print("Symmetric difference =", A.symmetric_difference(B))
```

**Q4(x)**

```python
print("A is a subset of B:", A.issubset(B))
print("B is a superset of A:", B.issuperset(A))
```

**Q4(xi)**

```python
val = int(input("Enter a value to remove from A: "))
A.discard(val)
print("A after discard =", A)

# discard() does nothing if the value is missing, remove() would raise a KeyError.
```

## Q5 - Dictionary

```python
my_dict = {
    "name": "Gurkirpa Singh Sarao",
    "roll_no": "1024170378",
    "branch": "Computer Science and Engineering",
    "age": 20,
    "city": "Patiala"
}
```

**Q5(i)**

```python
my_dict["location"] = my_dict.pop("city")   # pop returns the old value, so nothing is retyped
print(my_dict)
```

**Q5(ii)**

```python
my_dict["cgpa"] = 9.63
print(my_dict)
```

**Q5(iii)**

```python
my_dict["age"] = my_dict["age"] + 1
print(my_dict)
```

**Q5(iv)**

```python
d1 = my_dict.copy()
removed = d1.pop("branch")
print("pop() returned:", removed)
print(d1)

d2 = my_dict.copy()
del d2["branch"]
print("After del:", d2)

# pop() removes the key and returns its value, del just removes it and returns nothing.
```

**Q5(v)**

```python
for key, val in my_dict.items():
    print(key, "->", val)
```

**Q5(vi)**

```python
if "email" in my_dict:
    print(my_dict["email"])
else:
    print("Email key not present in the dictionary")
```

**Q5(vii)**

```python
friend_dict = {
    "name": "Karan",
    "roll_no": "1024170382",
    "branch": "Computer Science and Engineering",
    "age": 20,
    "city": "Ludhiana"
}

merged = {**my_dict, **friend_dict}
print("Merged:", merged)

# On a clash the dictionary written second wins, so friend_dict overwrites my_dict here.
```
