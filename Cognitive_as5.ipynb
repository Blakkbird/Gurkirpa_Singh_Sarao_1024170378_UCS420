## Ques 1

A 5 element array, then add 2, multiply by 3 and divide by 2. Each of these is a
vectorised operation, so numpy applies it to every element at once instead of me
looping. The original array is not touched since each line makes a new array.

```python
# Q1
arr = np.array([4, 8, 15, 16, 23])
print("original:", arr)

print("a) +2 :", arr + 2)
print("b) *3 :", arr * 3)
print("c) /2 :", arr / 2)

print("original is unchanged:", arr)
```

## Ques 2

Part (a) is just a reversed slice. For part (b) I count how many times each value
appears with `np.unique`, pick the value with the highest count, and then use `np.where` to get every position where it occurs.

```python
# Q2 a
arr = np.array([1, 2, 3, 6, 4, 5])
print("original:", arr)
print("reversed:", arr[::-1])
```

```python
# Q2 b
def most_frequent(a):
    vals, counts = np.unique(a, return_counts=True)
    val = vals[counts.argmax()]
    idx = np.where(a == val)[0]
    return val, idx


x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

for name, a in [("x", x), ("y", y)]:
    val, idx = most_frequent(a)
    print(name, "=", a)
    print("  most frequent:", val, "appearing", len(idx), "times")
    print("  at indices:", idx)
```

Note that y has a tie, 1 and 3 both come up three times. `argmax` returns the
first one it hits, so it reports 1. If all the tied values are wanted, then we can do:

```python
vals, counts = np.unique(y, return_counts=True)
tied = vals[counts == counts.max()]
print("values tied for most frequent in y:", tied)
for v in tied:
    print(" ", v, "at indices", np.where(y == v)[0])
```

## Ques 4

`linspace(10, 100, 25)` gives 25 evenly spaced values including both endpoints.
After printing the attributes I reshape it into a 25x1 column, which is what a
transpose of a row vector should look like.

```python
# Q4
gurkirpa = np.linspace(10, 100, 25)
print(gurkirpa)

print("dimensions :", gurkirpa.ndim)
print("shape      :", gurkirpa.shape)
print("size       :", gurkirpa.size)
print("dtype      :", gurkirpa.dtype)
print("bytes      :", gurkirpa.nbytes)
```

```python
# transpose using reshape
transposed = gurkirpa.reshape(25, 1)
print("shape after reshape:", transposed.shape)
print(transposed[:5])
```

Can the same be done with `T`? No. A 1-D array has only one axis, so there is
nothing for `T` to swap and it hands back the exact same array, still shape (25,).
It only works once the array actually has two axes.

```python
print("shape with T        :", gurkirpa.T.shape)
print("same object back    :", np.array_equal(gurkirpa, gurkirpa.T))
print("but T on the 2-D one:", transposed.T.shape)
```

## Ques 5

Twelve values in a 3x4 array, then the usual statistics. Reshape into 4x3 works
because 12 elements fit exactly. Resizing to 2x3 needs only 6, so `np.resize`
keeps the first six values in row order and drops the rest.
