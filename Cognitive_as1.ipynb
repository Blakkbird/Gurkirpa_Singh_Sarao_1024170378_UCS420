### Assignment 5.3

```python
n = int(input("Enter n: "))
total = 0

for i in range(2, n + 1):
    prime = True
    for j in range(2, int(i ** 0.5) + 1):
        if i % j == 0:
            prime = False
            break
    if prime:
        total += i

print("Sum of primes =", total)
```

### Assignment 6.1

```python
def add_odd(n):
    total = 0
    for i in range(1, n + 1):
        if i % 2 != 0:
            total += i
    return total


n = int(input("Enter n: "))
print("Sum of odd numbers =", add_odd(n))
```

### Assignment 6.2

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True


def add_prime(n):
    total = 0
    for i in range(2, n + 1):
        if is_prime(i):
            total += i
    return total


n = int(input("Enter n: "))
print("Sum of primes =", add_prime(n))
```
